# Global Terrorism Database Analysis

```python
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("globalterrorismdb_0718dist.csv", encoding="ISO-8859-1", low_memory=False)
```

```python
rename_dict = {
    "iyear": "Year",
    "imonth": "Month",
    "iday": "Day",
    "country_txt": "Country",
    "region_txt": "Region",
    "attacktype1_txt": "AttackType",
    "target1": "Target",
    "nkill": "Killed",
    "nwound": "Wounded",
    "summary": "Summary",
    "gname": "Group",
    "targtype1_txt": "Target_type",
    "weaptype1_txt": "Weapon_type"
}
```

```python
df.rename(columns={k: v for k, v in rename_dict.items() if k in df.columns}, inplace=True)
```

```python
killed_col = "Killed" if "Killed" in df.columns else "nkill"
wounded_col = "Wounded" if "Wounded" in df.columns else "nwound"
```

```python
df["Casualties"] = df[killed_col].fillna(0) + df[wounded_col].fillna(0)
```

```python
available_columns = ["Year", "Month", "Country", "Region", "AttackType", "Target", "Group", "Killed", "Wounded", "Casualties", "latitude", "longitude"]
```

```python
if "city" in df.columns:
    available_columns.append("city")
elif "City" in df.columns:
    available_columns.append("City")
```

```python
core_cols = [col for col in available_columns if col in df.columns]
df_cleaned = df[core_cols].dropna(subset=["Year", "Country", "Region"])
```

```python
df_cleaned.to_csv("terrorism_cleaned.csv", index=False)
```
